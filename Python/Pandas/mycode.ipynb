```python
import pandas as pd

df = pd.read_csv("retail_sales.csv")
print("Dataset loaded")

# .head() prints first 5 rows of the dataset
df.head()
```

```python
print("Dataset shape: ")
print(df.shape)

print("\ncol names: ")
print(df.columns)

# Jupyter automatically displays the value of the last expression in a cell, but not previous expressions unless you explicitly print them.
print("\nData types: ")
df.dtypes
```

```python
# selecting single col

# prints the col without the header
df["Order_ID"]
```

```python
# displays the col along with header
df[["Order_ID"]]

# displays the last 5 records
df[["Order_ID"]].tail()
```

```python
# selecting multiple cols
df[["Order_ID","Date","Customer","City"]].tail()
```

```python
print("filtering data with single filter: ")
electronics = df[df["Category"] == "Electronics"]
electronics.head()
```

```python
print("filtering data with multiple filter: ")

# when using multiple filters, enclose each filter in '()'
filtered_df = df[(df["Price"]>10000) & (df["Category"] == "Electronics")]

# to display the first 4 elements
filtered_df.head(4)
```

```python
# Adding new col
df["Total_Sales"] = df["Quantity"] * df["Price"]
df.head(2)
```

```python
# Sorting data

# ascending by default
# sorted_df = df.sort_values(by="Price")

# descending
sorted_df = df.sort_values(by="Price", ascending = False)
sorted_df.head(10)
```

```python
# Handling missing values
df.head().isnull()
```

```python
# unique values
df["City"].unique()
```

```python
# frequency
df["Category"].value_counts()
```

```python
# group by operation
grouped = df.groupby("Category")
max_sales = grouped["Total_Sales"].max()
max_sales
```

```python
# multiple aggregations
sales = df.groupby("City")["Total_Sales"].agg(["max", "min", "sum"])
sales
```

```python
# index = rows
# columns = columns
# values = numbers to calculate
# aggfunc = calculation to perform

pivot = pd.pivot_table(
    df,
    values="Total_Sales",
    index="City",
    columns="Category",
    aggfunc="sum"
)
pivot
```

```python
df["Date"] = pd.to_datetime(df["Date"]).dt.month
df["Date"]
```

```python
# export processed data

df.to_csv("Processed_sales.csv", index=False)
```
